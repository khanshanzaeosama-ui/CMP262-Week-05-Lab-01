### Task 13 - One-Condition Filter
Create a Boolean mask for workshops with a satisfaction score of 4.5 or higher. Use the mask to display the matching rows.


````python
# Write your code here
mask = workshops["Satisfaction"] >= 4.5
workshops[mask]
````

### Task 14 - Multiple-Condition Filter
Display workshops that are `In Person` **and** have at least 20 attendees. Use parentheses around each comparison and the pandas `&` operator.


````python
# Write your code here
workshops[(workshops["Delivery_Mode"] == "In Person") & (workshops["Attended"] >=20)]
````

### Task 15 - OR and NOT Filters
1. Display workshops whose category is `Programming` **or** `Data Science`.
2. Display workshops that are **not** online.
Use `|` for OR and `~` for NOT.


````python
# Write your code here
workshops[(workshops["Category"] == "Programming") | (workshops["Category"] == "Data Science")]
workshops[~(workshops["Delivery_Mode"] == "Online")]
````
